ALGUNS INSIGHTS

Taxa de execução financeira:

```python
aporte_total = movimentacao_financeira['aporte'].sum()
pagamento_total = movimentacao_financeira['pagamento'].sum()

execucao_financeira = (pagamento_total / aporte_total) * 100

print(f'Execução Financeira: {execucao_financeira:.2f}%')
```

Ranking dos fundos por saldo:

```python
ranking_saldo = (
    movimentacao_financeira.merge(fundo[['idfundo','nome_fundo']],on='idfundo')
    [['nome_fundo','saldo']].sort_values(by='saldo',ascending=False))

print(ranking_saldo)
```

Percentual de execução dos contratos:

```python
contrato['percentual_execucao'] = (contrato['valor_execucao'] / contrato['valor_contrato']) * 100

print(contrato[['idcontrato','percentual_execucao']])
```

Contratos críticos - execução menor que 60%:

```python
contratos_criticos = contrato[contrato['percentual_execucao'] < 60]

print(contratos_criticos)
```

Valor ainda não executado:

```python
contrato['saldo_a_executar'] = (contrato['valor_contrato'] - contrato['valor_execucao'])

print(contrato[['idcontrato','saldo_a_executar']])
```

Índice de divergência das conciliações:

```python
conciliacao['indice_divergencia']=(conciliacao['quantidade_divergencias']/conciliacao['quantidade_lancamentos'])*100

print(conciliacao[['idfundo','indice_divergencia']])
```

Prestação de contas pendentes:

```python
pendentes = prestacao_contas[prestacao_contas['data_aprovacao'].isna()]

print(pendentes)
```

Tempo médio de aprovação:

```python
prestacao_contas['data_envio'] = pd.to_datetime(prestacao_contas['data_envio'])

prestacao_contas['data_aprovacao'] = pd.to_datetime(prestacao_contas['data_aprovacao'])

prestacoes_aprovadas = prestacao_contas.dropna(subset=['data_aprovacao'])

prestacoes_aprovadas['dias_aprovacao'] = (prestacoes_aprovadas['data_aprovacao']-prestacoes_aprovadas['data_envio']
).dt.days

tempo_medio = (prestacoes_aprovadas['dias_aprovacao'].mean())

print(f'Tempo médio: {tempo_medio:.0f} dias')
```

Contratos próximos do vencimento:

```python
from datetime import datetime

hoje = pd.Timestamp.today()

contrato['data_fim'] = pd.to_datetime(contrato['data_fim'])

contrato['dias_para_vencer'] = (contrato['data_fim'] - hoje).dt.days

vencimento = contrato[contrato['dias_para_vencer'] <= 90]

print(vencimento)
```

Score de risco do fundo:

```python
base_risco = (contrato.groupby('idfundo')['percentual_execucao'].mean().reset_index())

base_risco = (base_risco.merge(conciliacao[['idfundo','quantidade_divergencias']],on='idfundo'))

def score(row):
    pontos = 0
    if row['percentual_execucao'] < 60:
        pontos += 2
    elif row['percentual_execucao'] < 80:
        pontos += 1
    if row['quantidade_divergencias'] >= 2:
        pontos += 2
    if pontos >= 3:
        return 'Alto'
    if pontos >= 2:
        return 'Médio'
    return 'Baixo'

base_risco['risco'] = (base_risco.apply(score, axis=1))

print(base_risco)
```

Contrato com maior taxa de execução e contrato com taxa de execução inferior a 50%:

```python
melhor = (contrato.sort_values(by='percentual_execucao',ascending=False).iloc[0])

pior = (contrato.sort_values(by='percentual_execucao').iloc[0])

print(f'''O contrato {melhor["idcontrato"]} apresenta a maior taxa de execução({melhor["percentual_execucao"]:.2f}%).

O contrato {pior["idcontrato"]} apresenta a menor taxa de execução({pior["percentual_execucao"]:.2f}%),
sendo recomendado acompanhamento prioritário.''')
```

Quais fundos tranformam aportes em pagamentos de forma mais eficiente:

```python
df = movimentacao_financeira.merge(fundo[['idfundo','nome_fundo']],on='idfundo')

df['execucao_%'] = (df['pagamento'] / df['aporte']) * 100

plt.figure(figsize=(10,5))

sns.barplot(data=df, x='nome_fundo', y='execucao_%', palette='Blues_r')
plt.title('Execução Financeira por Fundo')
plt.ylabel('% Execução')
plt.xlabel('')
plt.xticks(rotation=30)
plt.show()
```

Fundos com maior volume de recursos sob gestão:

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df.sort_values('saldo', ascending=False), x='saldo',y='nome_fundo',palette='Greens')
plt.title('Saldo por Fundo')
plt.xlabel('Saldo (R$)')
plt.ylabel('')
plt.show()
```

Contratos com baixa execução:

```python
contrato['percentual_execucao'] = (contrato['valor_execucao']/contrato['valor_contrato']) * 100

plt.figure(figsize=(10,5))
sns.barplot(data=contrato,x='idcontrato',y='percentual_execucao',palette='Oranges')
plt.axhline(60,color='red',linestyle='--',label='Limite crítico')
plt.legend()
plt.title('Percentual de Execução dos Contratos')
plt.ylabel('%')
plt.show()
```

Localizar os contratos mais críticos:

```python
contrato['saldo_a_executar'] = (contrato['valor_contrato']-contrato['valor_execucao'])
heat = contrato[['percentual_execucao','saldo_a_executar']]

sns.heatmap(heat.corr(),annot=True,cmap='Reds')
plt.title('Correlação de Indicadores')
plt.show()
```

Fundos com maior risco operacional:

```python
conciliacao = conciliacao.merge(fundo[['idfundo','nome_fundo']],on='idfundo')

plt.figure(figsize=(10,5))
sns.barplot(data=conciliacao,x='nome_fundo',y='quantidade_divergencias',palette='Reds')
plt.title('Divergências de Conciliação')
plt.ylabel('Quantidade')
plt.xticks(rotation=30)
plt.show()
```

Índice de divergência:

```python
conciliacao['indice_divergencia']=(conciliacao['quantidade_divergencias']/conciliacao['quantidade_lancamentos'])*100

plt.figure(figsize=(10,5))
sns.barplot(data=conciliacao,x='nome_fundo',y='indice_divergencia',palette='rocket')
plt.title('Índice de Divergência (%)')
plt.ylabel('%')
plt.show()
```

Tempo de aprovação das prestações:


```python
prestacao_contas['data_envio'] = pd.to_datetime(prestacao_contas['data_envio'])
prestacao_contas['data_aprovacao'] = pd.to_datetime(prestacao_contas['data_aprovacao'])
validos = prestacao_contas.dropna(subset=['data_aprovacao'])
validos['dias_aprovacao'] = (validos['data_aprovacao'] - validos['data_envio']).dt.days

plt.figure(figsize=(8,5))
sns.boxplot(y=validos['dias_aprovacao'],color='lightblue')
plt.title('Tempo de Aprovação das Prestações')
plt.ylabel('Dias')
plt.show()
```

Score de Risco:

```python
risco = contrato.groupby('idfundo').agg({'percentual_execucao':'mean'}).reset_index()
risco = risco.merge(conciliacao[['idfundo','quantidade_divergencias']],on='idfundo')
risco = risco.merge(fundo[['idfundo','nome_fundo']],on='idfundo')

plt.figure(figsize=(10,5))
sns.scatterplot(data=risco,x='percentual_execucao',y='quantidade_divergencias',hue='nome_fundo',s=300)
plt.title('Matriz de Risco dos Fundos')
plt.xlabel('Execução Financeira (%)')
plt.ylabel('Divergências')
plt.show()
```

Quadrante Superior Direito → Alto risco

Quadrante Inferior Direito → Boa execução

Quadrante Superior Esquerdo → Divergências elevadas

Quadrante Inferior Esquerdo → Necessita atenção imediata
